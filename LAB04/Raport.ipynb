## 1. Komanda `print`

Komanda `print()` përdoret për të shfaqur tekst ose rezultate në ekran.


````python
print("Hello, Python!")
print("Ky është laboratori 04.")
````

## 2. Variablat

Variablat përdoren për të ruajtur vlera.

Për shembull:


````python
a = 5
b = 3

print("a =", a)
print("b =", b)
````

## 3. Llogaritje të thjeshta

Në Python mund të bëjmë veprime matematikore si mbledhje, zbritje, shumëzim dhe pjesëtim.


````python
mbledhje = a + b
zbritje = a - b
shumezim = a * b
pjesetim = a / b

print("Mbledhja:", mbledhje)
print("Zbritja:", zbritje)
print("Shumëzimi:", shumezim)
print("Pjesëtimi:", pjesetim)
````

## 4. Listat

Lista përdoret për të ruajtur disa vlera në një vend.


````python
matje = [10, 12, 9, 11, 13]

print("Matjet janë:", matje)
print("Matja e parë:", matje[0])
print("Numri i matjeve:", len(matje))
````

## 5. Mesatarja e matjeve

Mund të llogarisim mesataren e një liste me matje.


````python
mesatarja = sum(matje) / len(matje)

print("Mesatarja e matjeve është:", mesatarja)
````

## 6. Përdorimi i NumPy

`NumPy` është bibliotekë që përdoret për llogaritje numerike.

Zakonisht importohet kështu:

```python
import numpy as np
```


````python
import numpy as np

matje_np = np.array([10, 12, 9, 11, 13])

print("Matjet me NumPy:", matje_np)
print("Mesatarja:", np.mean(matje_np))
print("Vlera minimale:", np.min(matje_np))
print("Vlera maksimale:", np.max(matje_np))
````

## 7. Krijimi i një grafiku

Për grafikë përdorim bibliotekën `matplotlib`.


````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 12, 9, 11, 13]

plt.plot(x, y, marker="o")
plt.xlabel("Numri i matjes")
plt.ylabel("Vlera")
plt.title("Grafik i thjeshtë i matjeve")
plt.grid(True)
plt.show()
````
