# LAB 07 — Përcaktimi i Dendësisë

## 1. Qëllimi i eksperimentit

Qëllimi i këtij eksperimenti është përcaktimi i dendësisë së një trupi me formë të rregullt drejtkëndore.  
Dendësia nuk matet drejtpërdrejt, por llogaritet duke matur masën dhe përmasat e trupit.

Formula kryesore është:

$$
\rho = \frac{m}{V}
$$

ku:

- $\rho$ është dendësia,
- $m$ është masa,
- $V$ është vëllimi i trupit.

---

## 2. Të dhënat eksperimentale

Objekti i marrë në shqyrtim ka formë drejtkëndore.  
Për këtë arsye, vëllimi i tij llogaritet me formulën:

$$
V = a \cdot b \cdot c
$$

ku:

- $a$ është gjatësia,
- $b$ është gjerësia,
- $c$ është lartësia.

Tabela e matjeve:

| Nr. | Masa $m$ (g) | Gjatësia $a$ (cm) | Gjerësia $b$ (cm) | Lartësia $c$ (cm) |
|---:|---:|---:|---:|---:|
| 1 | 48.62 | 5.12 | 2.98 | 1.54 |
| 2 | 48.58 | 5.11 | 2.99 | 1.55 |
| 3 | 48.65 | 5.13 | 2.98 | 1.54 |
| 4 | 48.60 | 5.12 | 3.00 | 1.53 |
| 5 | 48.63 | 5.11 | 2.99 | 1.54 |

---

## 3. Vlerat mesatare

Vlera mesatare e masës llogaritet me formulën:

$$
\bar{m} = \frac{m_1 + m_2 + m_3 + m_4 + m_5}{5}
$$

$$
\bar{m} = \frac{48.62 + 48.58 + 48.65 + 48.60 + 48.63}{5}
$$

$$
\bar{m} = 48.616 \text{ g}
$$

Në të njëjtën mënyrë llogariten edhe vlerat mesatare të përmasave:

$$
\bar{a} = 5.118 \text{ cm}
$$

$$
\bar{b} = 2.988 \text{ cm}
$$

$$
\bar{c} = 1.540 \text{ cm}
$$

---

## 4. Llogaritja e vëllimit

Për trupin drejtkëndor kemi:

$$
V = a \cdot b \cdot c
$$

Duke zëvendësuar vlerat mesatare:

$$
V = 5.118 \cdot 2.988 \cdot 1.540
$$

$$
V = 23.551 \text{ cm}^3
$$

Pra, vëllimi i trupit është:

$$
V = 23.551 \text{ cm}^3
$$

---

## 5. Llogaritja e dendësisë

Formula e dendësisë është:

$$
\rho = \frac{m}{V}
$$

Duke zëvendësuar vlerat:

$$
\rho = \frac{48.616}{23.551}
$$

$$
\rho = 2.064 \text{ g/cm}^3
$$

Pra, dendësia e trupit është:

$$
\rho = 2.064 \text{ g/cm}^3
$$

---

## 6. Llogaritja e pasigurive

Nga matjet eksperimentale marrim këto pasiguri:

$$
\Delta m = 0.027 \text{ g}
$$

$$
\Delta a = 0.008 \text{ cm}
$$

$$
\Delta b = 0.008 \text{ cm}
$$

$$
\Delta c = 0.007 \text{ cm}
$$

Pasiguria relative e vëllimit jepet nga formula:

$$
\frac{\Delta V}{V}
=
\frac{\Delta a}{a}
+
\frac{\Delta b}{b}
+
\frac{\Delta c}{c}
$$

Duke zëvendësuar vlerat:

$$
\frac{\Delta V}{V}
=
\frac{0.008}{5.118}
+
\frac{0.008}{2.988}
+
\frac{0.007}{1.540}
$$

$$
\frac{\Delta V}{V} = 0.00903
$$

Pasiguria absolute e vëllimit është:

$$
\Delta V = V \cdot 0.00903
$$

$$
\Delta V = 23.551 \cdot 0.00903
$$

$$
\Delta V = 0.213 \text{ cm}^3
$$

Prandaj:

$$
V = (23.551 \pm 0.213) \text{ cm}^3
$$

---

## 7. Pasiguria e dendësisë

Dendësia llogaritet me formulën:

$$
\rho = \frac{m}{V}
$$

Pasiguria relative e dendësisë jepet nga:

$$
\frac{\Delta \rho}{\rho}
=
\frac{\Delta m}{m}
+
\frac{\Delta V}{V}
$$

Duke zëvendësuar vlerat:

$$
\frac{\Delta \rho}{\rho}
=
\frac{0.027}{48.616}
+
0.00903
$$

$$
\frac{\Delta \rho}{\rho} = 0.00958
$$

Pasiguria absolute e dendësisë është:

$$
\Delta \rho = \rho \cdot 0.00958
$$

$$
\Delta \rho = 2.064 \cdot 0.00958
$$

$$
\Delta \rho = 0.020 \text{ g/cm}^3
$$

---

## 8. Rezultati përfundimtar

Rezultati përfundimtar i dendësisë është:

$$
\rho = (2.064 \pm 0.020) \text{ g/cm}^3
$$

Duke e rrumbullakosur:

$$
\boxed{\rho = (2.06 \pm 0.02) \text{ g/cm}^3}
$$

---

## 9. Përfundimi

Në këtë eksperiment u përcaktua dendësia e një trupi me formë drejtkëndore.  
Fillimisht u mat masa dhe përmasat e trupit disa herë. Më pas u llogaritën vlerat mesatare të masës, gjatësisë, gjerësisë dhe lartësisë.

Duke përdorur formulën:

$$
V = a \cdot b \cdot c
$$

u gjet vëllimi i trupit. Pastaj, me formulën:

$$
\rho = \frac{m}{V}
$$

u përcaktua dendësia e tij.

Rezultati përfundimtar doli:

$$
\rho = (2.06 \pm 0.02) \text{ g/cm}^3
$$

Pasiguria tregon se matjet eksperimentale nuk janë plotësisht të sakta, pasi çdo instrument matës ka kufizime. Megjithatë, rezultati është i pranueshëm dhe paraqet dendësinë mesatare të objektit të matur.
