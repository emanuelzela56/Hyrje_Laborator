# LAB 09 — Matja e Kohës së Ekzekutimit të një Algoritmi

## 1. Qëllimi i laboratorit

Qëllimi i këtij laboratori është të studiohet se si ndryshon koha e ekzekutimit të një algoritmi kur ndryshon madhësia e hyrjes $n$.

Në këtë laborator përdoret algoritmi i shumës së elementeve të një liste.

Për madhësi të ndryshme të listës, matet koha që i duhet programit për të mbledhur të gjithë elementet. Më pas ndërtohet grafiku log-log për të analizuar varësinë midis kohës së ekzekutimit dhe madhësisë së hyrjes.

---

## 2. Teoria

Koha e ekzekutimit të një algoritmi zakonisht varet nga madhësia e hyrjes $n$.

Në përgjithësi, kjo varësi mund të shkruhet:

$$
T(n) \sim n^p
$$

ku:

- $T(n)$ është koha e ekzekutimit;
- $n$ është madhësia e hyrjes;
- $p$ është fuqia që tregon mënyrën si rritet koha.

Për algoritmin e shumës së elementeve të një liste, programi kalon një herë nëpër të gjithë elementet e listës.

Prandaj, nëse lista ka $n$ elemente, algoritmi kryen afërsisht $n$ veprime.

Për këtë arsye pritet që koha e ekzekutimit të rritet në mënyrë lineare:

$$
T(n) \sim n
$$

Kompleksiteti teorik i këtij algoritmi është:

$$
O(n)
$$

---

## 3. Algoritmi i përdorur

Algoritmi i përdorur është shuma e elementeve të një liste.

Ideja e algoritmit është:

1. krijohet një variabël `total`;
2. programi kalon nëpër çdo element të listës;
3. çdo element shtohet te `total`;
4. në fund kthehet shuma totale.

Në formë të thjeshtë:

```python
total = 0

for x in data:
    total += x

````python

import time
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Funksion për matjen e kohës
# -----------------------------

def measure_time(function, data, repeats=5):
    times = []
    
    for _ in range(repeats):
        start = time.perf_counter()
        function(data)
        end = time.perf_counter()
        
        times.append(end - start)
    
    return np.mean(times)


# -----------------------------
# Algoritmi: Shuma e elementeve
# -----------------------------

def sum_algorithm(data):
    total = 0
    
    for x in data:
        total += x
    
    return total


# -----------------------------
# Madhësitë e hyrjes
# -----------------------------

n_values = np.array([
    100, 300, 1000, 3000,
    10000, 30000, 100000, 300000
])


# -----------------------------
# Matja e kohës
# -----------------------------

times = []

for n in n_values:
    data = np.random.randint(0, 100, size=n)
    T = measure_time(sum_algorithm, data, repeats=5)
    times.append(T)

times = np.array(times)


# -----------------------------
# Përshtatja log-log
# -----------------------------

coeff = np.polyfit(np.log(n_values), np.log(times), 1)

slope = coeff[0]
intercept = coeff[1]

fit_times = np.exp(intercept) * n_values**slope


# -----------------------------
# Printimi i rezultateve
# -----------------------------

print("Rezultatet për algoritmin e shumës:")
print("-----------------------------------")

for n, T in zip(n_values, times):
    print(f"n = {n:7d}, T = {T:.8f} s")

print("\nPjerrësia në grafikun log-log:")
print(f"p ≈ {slope:.2f}")

print("\nInterpretimi:")
if 0.8 <= slope <= 1.2:
    print("Algoritmi ka sjellje afërsisht lineare, pra T(n) ~ n.")
else:
    print("Pjerrësia nuk është shumë afër 1, për shkak të luhatjeve eksperimentale.")


# -----------------------------
# Grafiku normal
# -----------------------------

plt.figure()
plt.plot(n_values, times, marker="o")
plt.xlabel("Madhësia e hyrjes n")
plt.ylabel("Koha e ekzekutimit T(n) [s]")
plt.title("Koha e ekzekutimit kundrejt madhësisë së hyrjes")
plt.grid(True)
plt.show()


# -----------------------------
# Grafiku log-log
# -----------------------------

plt.figure()
plt.loglog(n_values, times, marker="o", label="Matjet eksperimentale")
plt.loglog(n_values, fit_times, linestyle="--", label=f"Përshtatja: p = {slope:.2f}")

plt.xlabel("Madhësia e hyrjes n")
plt.ylabel("Koha e ekzekutimit T(n) [s]")
plt.title("Grafiku log-log për algoritmin e shumës")
plt.grid(True, which="both")
plt.legend()
plt.show()
````
