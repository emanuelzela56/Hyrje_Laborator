# LAB 08 — Matja e Kohës së Reagimit

## 1. Qëllimi i laboratorit

Qëllimi i këtij laboratori është përcaktimi i kohës së reagimit të një personi duke përdorur metodën e rënies së vizores.

Në këtë eksperiment matet distanca që përshkon vizorja gjatë rënies së lirë, para se ajo të kapet nga personi. Më pas, duke përdorur formulat e lëvizjes me nxitim konstant, llogaritet koha e reagimit.

Koha e reagimit nuk është gjithmonë e njëjtë, sepse ajo ndikohet nga disa faktorë, si:

- përqendrimi;
- lodhja;
- vëmendja;
- parashikimi i momentit të rënies;
- koordinimi sy-dorë;
- gabimet eksperimentale gjatë leximit të vizores.

---

## 2. Teoria

Kur vizorja bie lirshëm, ajo kryen lëvizje me nxitim konstant nën veprimin e gravitetit.

Formula e rrugës për lëvizjen me nxitim konstant është:

$$
d = \frac{1}{2}gt^2
$$

ku:

- $d$ është distanca e rënies së vizores;
- $g$ është nxitimi gravitacional;
- $t$ është koha e reagimit.

Duke zgjidhur formulën sipas kohës, marrim:

$$
t = \sqrt{\frac{2d}{g}}
$$

Në këtë laborator marrim:

$$
g = 9.81 \text{ m/s}^2
$$

Meqë distanca në eksperiment matet në centimetra, ajo duhet të kthehet në metra përpara llogaritjes:

$$
d_{\text{m}} = \frac{d_{\text{cm}}}{100}
$$

## 4. Llogaritja e kohës së reagimit

Për çdo matje përdoret formula:

$$
t = \sqrt{\frac{2d}{g}}
$$

Për shembull, për matjen e parë kemi:

$$
d = 10 \text{ cm}
$$

Fillimisht e kthejmë në metra:

$$
d = \frac{10}{100} = 0.10 \text{ m}
$$

Pastaj llogarisim kohën:

$$
t = \sqrt{\frac{2 \cdot 0.10}{9.81}}
$$

$$
t = \sqrt{0.02039}
$$

$$
t \approx 0.143 \text{ s}
$$

E njëjta procedurë përdoret për të gjitha matjet e tjera.

---

## 5. Vlera mesatare

Koha mesatare e reagimit llogaritet me formulën:

$$
\bar{t} = \frac{1}{N}\sum_{i=1}^{N}t_i
$$

ku:

- $\bar{t}$ është koha mesatare e reagimit;
- $N$ është numri i matjeve;
- $t_i$ janë kohët individuale të reagimit.

Nga përpunimi i të dhënave merret:

$$
\bar{t} = 0.140 \text{ s}
$$

---

## 6. Devijimi standard

Devijimi standard tregon sa shumë ndryshojnë matjet nga vlera mesatare.

Formula e devijimit standard është:

$$
\sigma = \sqrt{\frac{1}{N-1}\sum_{i=1}^{N}(t_i-\bar{t})^2}
$$

ku:

- $\sigma$ është devijimi standard;
- $t_i$ janë kohët individuale;
- $\bar{t}$ është koha mesatare;
- $N$ është numri i matjeve.

Nga llogaritjet merret:

$$
\sigma = 0.025 \text{ s}
$$

---

## 7. Vlerat minimale dhe maksimale

Nga të dhënat e përpunuara, koha minimale dhe maksimale e reagimit janë:

$$
t_{\min} = 0.078 \text{ s}
$$

$$
t_{\max} = 0.181 \text{ s}
$$

Vlera minimale lidhet me matjen më të vogël të distancës, pra:

$$
d = 3 \text{ cm}
$$

Kjo matje është dukshëm më e vogël se matjet e tjera dhe mund të konsiderohet si një vlerë jo shumë e zakonshme, ose si një rezultat i ndikuar nga parashikimi i rënies së vizores.

---

## 8. Rezultati përfundimtar

Rezultati përfundimtar i kohës së reagimit shprehet në formën:

$$
t = (\bar{t} \pm \sigma)
$$

Duke zëvendësuar vlerat:

$$
t = (0.140 \pm 0.025) \text{ s}
$$

Duke e rrumbullakosur:

$$
\boxed{t = (0.14 \pm 0.03) \text{ s}}
$$

---

## 9. Histogrami dhe shpërndarja

Për të analizuar shpërndarjen e matjeve, ndërtohet histogrami i kohëve të reagimit.

Histogrami tregon se sa shpesh shfaqen vlera të ndryshme të kohës së reagimit. Nëse të dhënat janë të shpërndara rreth një vlere mesatare dhe kanë formë afërsisht kambane, ato mund të krahasohen me shpërndarjen normale ose Gaussiane.

Shpërndarja Gaussiane ka formën:

$$
f(t) = \frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{(t-\bar{t})^2}{2\sigma^2}}
$$

ku:

- $\bar{t}$ është vlera mesatare;
- $\sigma$ është devijimi standard;
- $t$ është koha e reagimit.

---

## 10. Diskutim

Nga matjet e kryera vihet re se koha e reagimit nuk është konstante. Kjo ndodh sepse reagimi i njeriut ndryshon nga një provë në tjetrën.

Në disa raste personi mund të jetë më i përqendruar dhe ta kapë vizoren më shpejt. Në raste të tjera, ai mund të reagojë më ngadalë për shkak të lodhjes, mungesës së vëmendjes ose vonesës në koordinimin sy-dorë.

Vlera mesatare e kohës së reagimit doli:

$$
\bar{t} = 0.140 \text{ s}
$$

Kjo tregon se personi reagoi mesatarisht për rreth 0.14 sekonda.

Devijimi standard doli:

$$
\sigma = 0.025 \text{ s}
$$

Kjo tregon se matjet kanë një shpërndarje të caktuar rreth vlerës mesatare.

Matja me distancë:

$$
d = 3 \text{ cm}
$$

dha kohën më të vogël të reagimit:

$$
t = 0.078 \text{ s}
$$

Kjo vlerë mund të jetë rezultat i një reagimi shumë të shpejtë, por gjithashtu mund të vijë nga parashikimi i momentit të lëshimit të vizores ose nga një gabim gjatë leximit të shkallës së vizores.

Prandaj, matjet e përsëritura janë të rëndësishme sepse na japin një vlerë më të besueshme mesatare dhe na lejojnë të vlerësojmë pasigurinë e eksperimentit.

## 11. Përfundimi

Në këtë laborator u përcaktua koha e reagimit duke përdorur metodën e rënies së vizores.

Fillimisht u mat distanca që përshkoi vizorja gjatë rënies së lirë. Pastaj distanca u kthye nga centimetra në metra dhe u përdor formula:

$$
t = \sqrt{\frac{2d}{g}}
$$

Për secilën matje u llogarit koha përkatëse e reagimit. Më pas u përcaktua vlera mesatare dhe devijimi standard.

Rezultati përfundimtar i eksperimentit është:

$$
\boxed{t = (0.14 \pm 0.03) \text{ s}}
$$

Ky rezultat tregon se koha mesatare e reagimit është rreth 0.14 sekonda. Ndryshimet ndërmjet matjeve tregojnë se reagimi njerëzor nuk është plotësisht konstant dhe se çdo matje eksperimentale shoqërohet me pasiguri.

```python
import numpy as np
import matplotlib.pyplot as plt

d_cm = np.array([
    10, 15, 12, 14, 9, 8, 9, 14, 9, 7,
    9, 16, 11, 9, 6, 16, 11, 12, 3, 12,
    12, 10, 11, 5, 6, 8, 5, 9, 9
])

g = 9.81  # m/s^2

d_m = d_cm / 100

t = np.sqrt((2 * d_m) / g)

t_mean = np.mean(t)
t_std = np.std(t, ddof=1)

t_min = np.min(t)
t_max = np.max(t)

print("Numri i matjeve:", len(d_cm))

print("\nKohët e reagimit në sekonda:")
print(np.round(t, 3))

print("\nRezultatet:")
print(f"Koha mesatare = {t_mean:.3f} s")
print(f"Devijimi standard = {t_std:.3f} s")
print(f"Koha minimale = {t_min:.3f} s")
print(f"Koha maksimale = {t_max:.3f} s")

print("\nRezultati përfundimtar:")
print(f"t = ({t_mean:.3f} ± {t_std:.3f}) s")

plt.figure()
plt.hist(t, bins=8, density=True, edgecolor="black")
plt.xlabel("Koha e reagimit t (s)")
plt.ylabel("Dendësia e probabilitetit")
plt.title("Histogrami i kohës së reagimit")
plt.grid(True)
plt.show()

x = np.linspace(t_min, t_max, 200)

gaussian = (1 / (t_std * np.sqrt(2 * np.pi))) * np.exp(
    -((x - t_mean) ** 2) / (2 * t_std ** 2)
)

plt.figure()
plt.hist(t, bins=8, density=True, edgecolor="black", alpha=0.7, label="Të dhënat")
plt.plot(x, gaussian, label="Shpërndarja Gaussiane")
plt.xlabel("Koha e reagimit t (s)")
plt.ylabel("Dendësia e probabilitetit")
plt.title("Krahasimi me shpërndarjen Gaussiane")
plt.grid(True)
plt.legend()
plt.show()
```
