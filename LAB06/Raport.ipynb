# LAB 06 — Matja e Madhësive Bazë: Gjatësia, Masa dhe Koha

## 1. Qëllimi i laboratorit

Qëllimi i këtij laboratori është të kuptohet si kryhen matjet eksperimentale të madhësive bazë fizike, si gjatësia, masa dhe koha.

Në këtë laborator studiohen:

- matja e gjatësisë;
- matja e masës;
- matja e kohës;
- pasiguria e matjeve;
- gabimet sistematike dhe rastësore;
- raportimi i rezultatit përfundimtar.

Çdo matje fizike nuk është plotësisht e saktë, prandaj rezultati duhet të shprehet në formën:

$$
x = x_{\text{mat}} \pm \Delta x
$$

ku:

- $x_{\text{mat}}$ është vlera e matur;
- $\Delta x$ është pasiguria e matjes.

---

## 2. Teoria

Matja është procesi i krahasimit të një madhësie fizike me një njësi standarde.

Për disa matje të përsëritura, vlera mesatare llogaritet me formulën:

$$
\bar{x} = \frac{1}{N}\sum_{i=1}^{N}x_i
$$

ku:

- $\bar{x}$ është vlera mesatare;
- $N$ është numri i matjeve;
- $x_i$ janë matjet individuale.

Pasiguria mund të vijë nga:

1. **Instrumenti matës**  
   Për shembull, nëse vizorja ka ndarje më të vogël 1 mm, pasiguria merret afërsisht gjysma e ndarjes më të vogël.

2. **Gabimet rastësore**  
   Këto ndodhin për shkak të ndryshimeve të vogla gjatë matjeve të përsëritura.

3. **Gabimet sistematike**  
   Këto ndodhin kur instrumenti ka një zhvendosje konstante ose nuk është i kalibruar saktë.

---

# Pjesa A — Matja e Gjatësisë

## 3. Të dhënat eksperimentale për gjatësinë

U mat gjatësia e një objekti me tre instrumente të ndryshme:

- vizore;
- kalibër;
- mikrometër.

Tabela e matjeve:

| Nr. | Vizore (cm) | Kalibër (cm) | Mikrometër (cm) |
|---:|---:|---:|---:|
| 1 | 12.4 | 12.36 | 12.34 |
| 2 | 12.5 | 12.35 | 12.35 |
| 3 | 12.4 | 12.36 | 12.34 |
| 4 | 12.3 | 12.37 | 12.34 |
| 5 | 12.4 | 12.36 | 12.35 |

---

## 4. Llogaritja e gjatësisë mesatare

Për vizoren:

$$
\bar{x}_{v} = \frac{12.4 + 12.5 + 12.4 + 12.3 + 12.4}{5}
$$

$$
\bar{x}_{v} = 12.40 \text{ cm}
$$

Për kalibrin:

$$
\bar{x}_{k} = \frac{12.36 + 12.35 + 12.36 + 12.37 + 12.36}{5}
$$

$$
\bar{x}_{k} = 12.36 \text{ cm}
$$

Për mikrometrin:

$$
\bar{x}_{m} = \frac{12.34 + 12.35 + 12.34 + 12.34 + 12.35}{5}
$$

$$
\bar{x}_{m} = 12.344 \text{ cm}
$$

Duke krahasuar instrumentet, mikrometri jep matje më të sakta sepse ka precizion më të lartë.

# Pjesa B — Matja e Masës

## 5. Të dhënat eksperimentale për masën

Masa e të njëjtit objekt u mat 5 herë me peshore dixhitale.

| Nr. | Masa $m$ (g) |
|---:|---:|
| 1 | 48.62 |
| 2 | 48.61 |
| 3 | 48.63 |
| 4 | 48.60 |
| 5 | 48.62 |

Masa mesatare:

$$
\bar{m} = \frac{48.62 + 48.61 + 48.63 + 48.60 + 48.62}{5}
$$

$$
\bar{m} = 48.616 \text{ g}
$$

Pasiguria merret afërsisht:

$$
\Delta m = 0.01 \text{ g}
$$

Prandaj rezultati për masën është:

$$
m = (48.62 \pm 0.01) \text{ g}
$$

# Pjesa C — Matja e Kohës

## 6. Të dhënat eksperimentale për kohën

U mat koha e një procesi të thjeshtë duke përdorur kronometër. Matja u përsërit 10 herë.

| Nr. | Koha $t$ (s) |
|---:|---:|
| 1 | 1.42 |
| 2 | 1.39 |
| 3 | 1.44 |
| 4 | 1.41 |
| 5 | 1.40 |
| 6 | 1.43 |
| 7 | 1.38 |
| 8 | 1.42 |
| 9 | 1.41 |
| 10 | 1.40 |

Koha mesatare llogaritet me formulën:

$$
\bar{t} = \frac{1}{N}\sum_{i=1}^{N}t_i
$$

Duke zëvendësuar vlerat:

$$
\bar{t} = 1.41 \text{ s}
$$

Pasiguria e përafërt është:

$$
\Delta t = 0.02 \text{ s}
$$

Prandaj rezultati për kohën është:

$$
t = (1.41 \pm 0.02) \text{ s}
$$

---

## 7. Diskutim

Nga matjet e gjatësisë shihet se instrumentet e ndryshme japin rezultate pak të ndryshme. Kjo ndodh sepse çdo instrument ka rezolucion dhe precizion të ndryshëm.

Vizorja ka pasiguri më të madhe sepse ndarja më e vogël është më e madhe. Kalibri dhe mikrometri japin rezultate më të sakta, sepse mund të matin me më shumë shifra dhjetore.

Në matjen e masës, vlerat ishin shumë afër njëra-tjetrës. Kjo tregon se peshorja ishte relativisht e qëndrueshme dhe se gabimi rastësor ishte i vogël.

Në matjen e kohës, ndryshimet ishin më të mëdha për shkak të kohës së reagimit njerëzor. Kur përdoret kronometri manual, fillimi dhe ndalimi i matjes varen nga reagimi i personit që kryen eksperimentin.

Burimet kryesore të gabimeve janë:

- gabimet instrumentale;
- gabimet rastësore;
- gabimet nga koha e reagimit;
- gabimet nga leximi i gabuar i instrumentit.

---

$$

## 9. Përfundimi

Në këtë laborator u kryen matje të madhësive bazë fizike: gjatësisë, masës dhe kohës. Për secilën madhësi u bënë matje të përsëritura dhe u llogaritën vlerat mesatare.

U vu re se çdo matje ka pasiguri, e cila varet nga instrumenti dhe nga mënyra si kryhet matja. Instrumentet më precize, si kalibri dhe mikrometri, japin rezultate më të sakta sesa vizorja.

Matja e kohës kishte pasiguri më të madhe për shkak të kohës së reagimit njerëzor. Për këtë arsye, matjet e përsëritura janë të rëndësishme për të marrë një rezultat më të besueshëm.

Rezultati kryesor i laboratorit është se matja nuk është vetëm leximi i një numri, por një proces fizik që kërkon analizë, krahasim dhe vlerësim të pasigurisë.
