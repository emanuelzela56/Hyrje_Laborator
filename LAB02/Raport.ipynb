## 1. Çfarë është Git?

**Git** është një sistem për kontrollin e versioneve. Ai ruan historinë e ndryshimeve të skedarëve të një projekti.

Git përdoret për:

- zhvillim software
- analiza të të dhënave
- projekte kërkimore
- bashkëpunim shkencor
- dorëzim dhe organizim të raporteve laboratorike

Me Git mund të shohim çfarë kemi ndryshuar, të rikthehemi në versione të mëparshme dhe të punojmë më mirë në projekte.


## 2. Çfarë është GitHub?

**GitHub** është një platformë online ku ruhen repository-t Git.  
Në këtë kurs GitHub përdoret për të dorëzuar raportet laboratorike.

GitHub na lejon të:

- ruajmë projektin online
- dorëzojmë raportin te instruktori
- shohim historinë e commit-eve
- organizojmë notebook, të dhëna dhe figura

## 8. Workflow minimal i Git

Workflow kryesor që përdoret në laborator është:

```bash
git status
git add .
git commit -m "mesazh konciz cfare ndryshimesh keni bere"
git push
```

Kuptimi i komandave:

| Komanda | Kuptimi |
|---|---|
| `git status` | Tregon çfarë ka ndryshuar në projekt |
| `git add .` | Përgatit të gjitha ndryshimet për commit |
| `git commit -m "mesazh"` | Ruan një version të projektit |
| `git push` | Ngarkon ndryshimet në GitHub |


````python
# Workflow minimal i Git

workflow = {
    "git status": "Kontrollon gjendjen e projektit",
    "git add .": "Shton ndryshimet në staging area",
    'git commit -m "Update lab report"': "Krijon një version të ruajtur të punës",
    "git push": "Ngarkon ndryshimet në GitHub"
}

for command, explanation in workflow.items():
    print(f"{command:40} -> {explanation}")
````

## 9. Lidhja e repository lokal me GitHub

Pasi krijohet repository në GitHub, projekti lokal lidhet me të në këtë mënyrë:

```bash
git remote add origin https://github.com/USERNAME/REPOSITORY.git
git branch -M main
git push -u origin main
```

Këtu:

- `origin` është emri standard i repository online
- `main` është dega kryesore e projektit
- `push -u origin main` ngarkon projektin për herë të parë në GitHub


````python
# Shembull me placeholders
# USERNAME dhe REPOSITORY zëvendësohen me të dhënat reale të GitHub-it.

username = "USERNAME"
repository = "REPOSITORY"

remote_command = f"git remote add origin https://github.com/{username}/{repository}.git"

print(remote_command)
print("git branch -M main")
print("git push -u origin main")
````

## 11. Komandat minimale që duhen mbajtur mend

```bash

```

Komanda shtesë të dobishme:

```bash

```

- `git log` tregon historinë e commit-eve.
- `git log --oneline` e tregon historinë më shkurt.


````python
# Cheat sheet i komandave kryesore

commands = [
    "git init",
    "git status",
    "git add .",
    'git commit -m "mesazh"',
    "git push",
    "git pull",
    "git log",
    "git log --oneline"
]

for i, command in enumerate(commands, start=1):
    print(f"{i}. {command}")
````
